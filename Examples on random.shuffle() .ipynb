```python
#Shuffle()
import numpy as np
a=np.arange(10,20)
print("Original Values:\n{} Type:{} id:{} ".format(a,type(a),id(a)))
np.random.shuffle(a)
print("Shuffled Values:\n{}  Type:{} id:{} ".format(a,type(a),id(a)))
print(a.shape,a.ndim)
```

```python
import numpy as np
a=np.arange(10,20,2)
print("Original Values:\n{} Type:{} id:{} ".format(a,type(a),id(a)))
print("-"*50)
for i in range(1,6):
    np.random.shuffle(a)
    print("Shuffled Values:{}  Type:{} id:{} ".format(a,type(a),id(a)))
print("-"*50)    
print(a.shape,a.ndim)
```
