```python
#Create an object of nd array by eye()
import numpy as np
c=np.eye(4)
print("Content Of C:\n{}".format(c))
```

```python
#Create an object of nd array by eye()
import numpy as np
c=np.eye(3)
print("Content Of C:\n{}".format(c))
print("Dimensions :{}".format(c.ndim))
```

```python
#Create an object of nd array by eye()
import numpy as np
c=np.eye(4,dtype=int)
print("Content Of C:\n{}".format(c))
print("Dimensions :{}".format(c.ndim))
```

```python
#Create an object of nd array by eye()
import numpy as np
c=np.eye((4))
print("Content Of C:\n{}".format(c))
```
