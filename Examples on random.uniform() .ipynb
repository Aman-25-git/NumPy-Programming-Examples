```python
#5. uniform()  :  Uniform Distribution
#------------------------------------------------------------------------
#This Function Generates random sample(s) between Two Intervals(Low,High) According Programmer Choice.
#The Default random sample is low is 0.0 and high is 1.0
#Syntax-1:      varname=numpy.random.uniform()
#Syntax-2:	   varname=numpy.random.uniform(Low,High)	
#Syntax-3:	   varname=numpy.random.uniform(Low,High,size)	
```

```python
import numpy as np
a=np.random.uniform()
print("Random Values:{} Type:{} ".format(a,type(a)))
```

```python
a=np.random.uniform(10)
print("Random Values:{} Type:{} ".format(a,type(a)))
```

```python
a=np.random.uniform(10,20)
print("Random Values:{} Type:{} ".format(a,type(a)))
```

```python
a=np.random.uniform(10,20,5)#--->1D
print("Random Values:{} Type:{} ".format(a,type(a)))
print(a.shape,a.ndim)
```

```python
a=np.random.uniform(10,20,(2,3))#--->2D
print("Random Values:\n{} Type:{} ".format(a,type(a)))
print(a.shape,a.ndim)
```

```python
a=np.random.uniform(10,20,(2,2,3))#--->3D
print("Random Values:\n{} Type:{} ".format(a,type(a)))
print(a.shape,a.ndim)
```
