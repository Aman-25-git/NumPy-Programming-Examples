```python
#Complex Math operation by Traditional list object
lst=[1,2,3,4,5]
lst2=[8,7,6,5,4]
lst3=lst+lst2
print(lst3)
```

```python
#Complex Math operation by Traditional list object
lst=[1,2,3,4,5]
lst2=[8,7,6,5,4]
lst3=[]
for a,b in zip(lst,lst2):
    lst3.append(a+b)
print(lst)
print(lst2)
print(lst3)
```

```python
#Complex Math operation by object of ndarray
lst=[1,2,3,4,5]
lst2=[8,7,6,5,4]
a=an.array(lst)
b=an.array(lst2)
c=a+b
print(a)
print(b)
print(c)
```
