```python
#Creating an object of ndarray() by random 
#syntax:var=numpy.random.randint(val) -----------> Random Number from 0 to Value-1
import numpy.random as np
ra=np.randint(10)
print("Random Value is :{} Type :{} ".format(ra,type(ra)))
```

```python
#Syntax-3: var=numpy.random.randint(low,high,size)-------> \
#generates the random values between low to high with in key size if key size is 5 then it generates 5 random values type=ndarray()
```

```python
ra=np.randint(10000,50000,6)
print("{} Type :{} ".format(ra,type(ra)))
print("Dimension:{}".format(ra.ndim))
print("Shape:{}".format(ra.shape))
```

```python
ra=np.randint(10000,50000,(3,3))
print("{} Type :{} ".format(ra,type(ra)))
print("Dimension:{}".format(ra.ndim))
print("Shape:{}".format(ra.shape))
```

```python
ra=np.randint(10000,50000,(2,2,2))
print("{} Type :{} ".format(ra,type(ra)))
print("Dimension:{}".format(ra.ndim))
print("Shape:{}".format(ra.shape))
```

```python
ra=np.randint(10000,50000,(2,2,2,2))
print("{} Type :{} ".format(ra,type(ra)))
print("Dimension:{}".format(ra.ndim))
print("Shape:{}".format(ra.shape))
```

```python
ra=np.randint(10000,50000,(3,2))
print("{} Type :{} ".format(ra,type(ra)))
print("Dimension:{}".format(ra.ndim))
print("Shape:{}".format(ra.shape))
print("Number of Rows:{}".format(ra.shape[0]))
print("Number of Cols:{}".format(ra.shape[1]))
```

```python
ra=np.randint(10000,50000,(2,3,2))
print("{} Type :{} ".format(ra,type(ra)))
print("Dimension:{}".format(ra.ndim))
print("Shape:{}".format(ra.shape))
print("Number of Matrix:{}".format(ra.shape[0]))
print("Number of Rows:{}".format(ra.shape[1]))
print("Number of Cols:{}".format(ra.shape[2]))
```
