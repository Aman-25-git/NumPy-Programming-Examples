```python
#Create an obj of ndrray()
import numpy as np
a=np.array([1,2,3,4,5])
print("A Content:",a)
print("Address:",id(a))
```

```python
#Copy the Content of ndarray object
b=np.copy(a)
print("Copied content:{} Address:{}".format(b,id(b)))
```

```python
#Do the Changes in Original Object and Copied Object
a[1]=56
b[2]=45
print("A Content:",a)
print("Address:",id(a))
print("Copied content:{} Address:{}".format(b,id(b)))
```

```python
#Check the Base for Copied Object
print("Base Of A:",a.base)
```

```python
#Check the Base for ndarray Object
print("Base Of B:",b.base)
```

```python
#Alternatively
b1=a.copy()
print("B1 Content:",b1)
```

```python
print("Ndarray Original Object a={} id={}".format(a,id(a)))
print("Ndarray Copied Object b1={} id={}".format(b1,id(b1)))
```

```python
#Do the Changes in Original Object and Copied Object
a[0]=11
b1[0]=22
```

```python
print("Ndarray Original Object a={} id={}".format(a,id(a)))
print("Ndarray Copied Object b1={} id={}".format(b1,id(b1)))
```

```python
#Implementation of view()
#Create ndarray object of ndarray
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("Ndarray Original Object a={} id={}".format(a,id(a)))
```

```python
# view the Orginal Ndarry Object 'a'
b=np.view(a)#----AttributeError: module 'numpy' has no attribute 'view'
```

```python
# view the Orginal Ndarry Object 'a'
b=a.view()  # here view() Present in ndarray object
print("Ndarray View Object b={} id={}".format(b,id(b)))
```

```python
a[0]=23
b[2]=78
print("Ndarray Original Object a={} id={}".format(a,id(a)))
print("Ndarray View Object b={} id={}".format(b,id(b)))
```

```python
#Check the Base for Original Object
print("Base for Original Object=",a.base)
```

```python
#Check the Base for viewed  Object
print("Base for Viewed Object=",b.base)
```
