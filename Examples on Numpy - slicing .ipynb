```python
#Numpy--Slicing Operations on 1-D
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90])
print("Content of ndarray")
print(a)
print("Dim of a=",a.ndim)
print("Shape of a=",a.shape)
```

```python
#get 30,40,50 from a
print(a[2:5])
```

```python
#Get the Elements 60 70 80 90
print(a[5:])
```

```python
#Get the Elements 30,40,50,60 with -VE Slicing
print(a[-7:-3])
```

```python
#Get 10 30 50 70 90 from 2D Array
print(a[::2])
```

```python
#Get 20 40 60 80 from 2D Array
print(a[1::2])
```

```python
#Get 10 40 70 from 2DArray
print(a[::3])
```

```python
print(a)
```

```python
#Get 90,80,70,60 in Backward Direction from 2DArray
print(a[-1:-5:-1])
```

```python
#Get 90 70 50 30  10 in Backward Direction with -VE Step
print(a[-1::-2])
```

```python
#Get 90 70 50 30  10 in Backward Direction with +VE Step
print(a[::2][::-1])
```

```python
#Numpy--Slicing Operations on 2-D
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90,25,35,45,55,77,88,99]).reshape(4,4)
print("Content of ndarray")
print(a)
print("Dim of a=",a.ndim)
print("Shape of a=",a.shape)
```

```python
#Get [60,70,80] from 2D Array
print(a[1:2,1:])
```

```python
#Get [60,70,80] from 2D Array
print(a[1:2,1:4])
```

```python
#Get 60,70,25,35 from 2D Array
print(a[1:3,1:3])
```
