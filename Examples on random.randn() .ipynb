```python
#Syntax-1: var=numpy.random.randn()---->Generates random Floating Point Number 
import numpy.random as np
ra=np.randn()
print("Random Value:{} Type:{}".format(ra,type(ra)))
```

```python
ra=np.randn()
print("Random Value:{} Type:{}".format(ra,type(ra)))
```

```python
for i in range(1,6):
    ra=np.randn()
    print("Random Value:{} Type:{}".format(ra,type(ra)))
```

```python
#Syntax-2:varname=numpy.random.randn(Value)
#Gives  Number of Random Floating Point Values Depends on 'Value'-- +VE OR -VE---1D
```

```python
ra=np.randn(5)
print("Random Value:{} Type:{}".format(ra,type(ra)))
```

```python
ra=np.randn(5)
print("Random Value:{} Type:{}".format(ra,type(ra)))
print("\nDim={}  Shape={}".format(ra.ndim,ra.shape))
```

```python
#Syntax-3: varname=numpy.random.randn(Rows,Cols)
#Gives  Number of Random Floating Point Values Depends on 'Rows x Cols'-- +VE OR -VE---2D
```

```python
ra=np.randn(3,3)
print("Random Value\n:{} Type:{}".format(ra,type(ra)))
print("\nDim={}  Shape={}".format(ra.ndim,ra.shape))
```

```python
ra=np.randn(5,5)
print("Random Value\n:{} Type:{}".format(ra,type(ra)))
print("\nDim={}  Shape={}".format(ra.ndim,ra.shape))
```

```python
#Syntax-4:varname=numpy.random.randn(Matrices,Rows,Cols)
#Gives  Number of Random Floating Point Values Depends on 'Matrices x Rows x Cols'-- +VE OR -VE---3D
```

```python
ra=np.randn(2,2,2)
print("Random Value:{} Type:{}".format(ra,type(ra)))
print("\nDim={}  Shape={}".format(ra.ndim,ra.shape))
```

```python
ra=np.randn(3,2,2)
print("Random Value:{} Type:{}".format(ra,type(ra)))
print("\nDim={}  Shape={}".format(ra.ndim,ra.shape))
```
