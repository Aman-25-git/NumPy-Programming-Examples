```python
#rearrange the ndarray obj elements with permutation()
import numpy as np
a=np.random.randint(10,20,5)
print("Original content of a Before permutation():{}".format(a))
ra=np.random.permutation(a)
print("Original content of a After permutation():{}".format(a))
print("content of ra permutation():{}".format(ra))
```

```python
#Re-arrange the Elements of ndarray by using permutation()
import numpy.random as r
a=r.randint(10,21,size=7)
print("Original Content of ndarray a before permutation()={}".format(a))
for i in range(1,len(a)+1):
    ra=r.permutation(a)
    print("Time:{}-Re-Arranged Elements={} id={}".format(i,ra,id(ra)))
```
