```python
#Syntax:2:numpy.random.rand(value)
```

```python
a=np.rand(5) #5 random values are going to generate 
print("Random Value:{} Type:{} ".format(a,type(a)))
```

```python
a=np.rand(10)#10 random values are going to generate ------>1D
print("Random Value:{} Type:{} ".format(a,type(a)))
print(a.ndim)
```

```python
#Syntax:3:numpy.random.rand(rows,cols)
```

```python
a=np.rand(3,2)#6 random values are going to generate ----->2D
print("Random Value:{} Type:{} ".format(a,type(a)))
print(a.shape)
print(a.ndim)
```

```python
#Syntax:4:numpy.random.rand(Matrices,rows,cols)
```

```python
import numpy.random as np
a=np.rand(2,2,2)#6 random values are going to generate-->3D
print("Random Value:\n{} Type:{} ".format(a,type(a)))
print(a.shape)
print(a.ndim)
```

```python
for v in a:
    print("%0.2f"%v)
```
