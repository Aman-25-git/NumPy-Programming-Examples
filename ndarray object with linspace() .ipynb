```python
#Syntax:  varname=numpy.linspace(start,stop,num=50,endpoint=True,retstep=False,dtype)
import numpy as np
a=np.linspace(0,10)
print("Content of a\n{}  Type of a={}".format(a,type(a)))
print("Dim of a={}".format(a.ndim))
```
