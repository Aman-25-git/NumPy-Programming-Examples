```python
#Create an ndarray object of 2D Matrix identity
import numpy as np
lst=[0,1,1,0]
a=np.array(lst).reshape(2,2)
print("Content of a:\n{} Type:{}".format(a,type(a)))
print("-"*50)
print("Dimensions of a:{}".format(a.ndim))
```

```python
#create an object of ndarray ny using identity()
a=np.identity(2)
print("Content of a:\n{} Type:{}".format(a,type(a)))
print("-"*50)
print("Dimensions of a:{}".format(a.ndim))
```

```python
#create an object of ndarray ny using identity()
a=np.identity(2,int)
print("Content of a:\n{} Type:{}".format(a,type(a)))
print("-"*50)
print("Dimensions of a:{}".format(a.ndim))
```
