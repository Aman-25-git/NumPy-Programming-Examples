```python
#Code for Building  a Matrix with dim of 2 with All entries as 0
import numpy as np
lst=[0,0,0,0]
a=np.array(lst).reshape(2,2)
print("Zero Matrix")
print(a)
print("Dim of a=",a.ndim)
```

```python
#Code for Building  a Matrix with dim of 2 with All entries as 0
import numpy as np
lst=[[0,0,0],[0,0,0],[0,0,0]]
a=np.array(lst)
print("Zero Matrix")
print(a)
print("Dim of a=",a.ndim)
```

```python
#Creating an ndarray object with zeros()
#All Entries with 8 Zero's
import numpy as np
a=np.zeros(8)
print("Content of a\n{} Type:{} ".format(a,type(a)))
print("-"*50)
print("Dimensions of a:{}".format(a.ndim))
```
