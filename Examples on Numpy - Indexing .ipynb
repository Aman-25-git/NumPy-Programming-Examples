```python
#Numpy--Indexing Operation
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80])
print("Content of a:{}".format(a))
print("Dimension is:{}".format(a.ndim))
```

```python
#Indexing Operation  on 1D
```

```python
print(a[0])
```

```python
print(a[-1])
```

```python
print(a[4])
```

```python
print(a[-4])
```

```python
print(a[7])
```

```python
print(a[len(a)-1])
```

```python
print(a[-len(a)])
```

```python
#get the Element 10 by using -ve Index
print(a[-a.size])
```

```python
#Indexing Operation  on 2D
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("Content of a:\n{}".format(a))
print("Dimension is:{}".format(a.ndim))
```

```python
#Get the value called 50
print(a[0])
print(a[1,1])
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(9,1)
print("Content of a:\n{}".format(a))
print("Dimension is:{}".format(a.ndim))
print("Shape:{}".format(a.shape))
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(9,)
print("Content of a:\n{}".format(a))
print("Dimension is:{}".format(a.ndim))
print("Shape:{}".format(a.shape))
```

```python
import numpy as np
a=np.array([10,20,30,40,50,60,70,80,90]).reshape(3,3)
print("Content of a:\n{}".format(a))
print("Dimension is:{}".format(a.ndim))
print("Shape:{}".format(a.shape))
```

```python
#Get the value called 10
print(a[0,0])
```

```python
#Get the value called 90
print(a[-1][-1])
```

```python
print(a[1,1])
```

```python
print(a[2,0])
print(a[-1,-3])
```

```python
#Indexing Operation  on 3D
import numpy as np
a=np.arange(3*3*3).reshape(3,3,3)
print("Content of ndarray")
print(a)
print("Dim of a=",a.ndim)
print("Shape of a=",a.shape)
```

```python
print(a[1,1,1])
```

```python
print(a[-2,-2,-2])
```

```python
print(a[-1])
```

```python
print(a[1])
```

```python
#Get 13 from 3D Array
print(a[1][1,1])
```

```python
#Get 26 from 3D Array
print(a[-1,-1][-1])
```
