```python
#Vector Based Operations
lst=[10,20,30,40]
print("Content of lst={}  Type={}".format(lst,type(lst)))
for i in range(len(lst)):
    lst[i]=lst[i]+1
print("Content of lst={}  Type={}".format(lst,type(lst)))    
```
